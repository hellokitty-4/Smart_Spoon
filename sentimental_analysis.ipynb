```python
import pandas as pd


# Load your CSV
df = pd.read_csv("sentiment_responses.csv")

# Load BERT sentiment pipeline
from transformers import pipeline

# Use PyTorch explicitly
classifier = pipeline("sentiment-analysis", framework="pt")


# Only analyze textual responses (Q13 and Q14)
textual_questions = ["Q6","Q7","Q9","Q10", "Q13", "Q14", "Q16"]
results = []

for _, row in df.iterrows():
    for q in textual_questions:
        text = str(row[q])
        sentiment = classifier(text)[0]
        results.append({
            "Question": q,
            "Response": text,
            "Label": sentiment["label"],  # 'POSITIVE' or 'NEGATIVE'
            "Score": sentiment["score"]
        })

# Save the result
bert_df = pd.DataFrame(results)
bert_df.to_csv("bert_sentiment_results.csv", index=False)
```
